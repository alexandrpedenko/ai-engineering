```python
from dotenv import load_dotenv
from anthropic import Anthropic

load_dotenv()

client = Anthropic()
```

```python
model="claude-sonnet-4-6"
response = client.messages.create(
    model=model,
    max_tokens=300,
    messages=[
        {"role": "user", "content": "Hello, Claude!"}
    ]
)

for block in response.content:
    if block.type == "text":
        print(block.text)
```
