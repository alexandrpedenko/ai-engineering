```python
import pandas as pd

from notebook_utils import get_data_path

path = get_data_path("slow_responses.csv")

df = pd.read_csv(path, parse_dates=['Date'])
df["Date"] = df["Date"].dt.strftime("%Y-%m-%d %H:%M:%S")
df.head()
```

```python
# Export to Excel
df.to_excel('data.xlsx', index=False)
print("Data exported to data.xlsx")
```

```python
# Export to json
df.to_json('data.json', orient="columns", date_format="iso")
df.to_json('slow_responses.jsonl', orient="records", lines=True, date_format="iso")
```

```python
# Export to SQL 
import sqlite3

connection = sqlite3.connect('chat_data.db')
df.to_sql('chat_data.db', connection, if_exists='replace')
```

```python
# Export to YAML
import yaml

data_dict = df.to_dict(orient='records')

with open('chat_data.yaml', 'w') as file:
    yaml.dump(data_dict, file)
    print('Data exported to chat_data.yaml')
```
