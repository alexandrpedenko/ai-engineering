```python
import pandas as pd

data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'Boris', 'Eugene'],
    'Age': [25, 30, 35, 16, 45],
    'City': ['New York', 'Los Angeles', 'Chicago', 'Boston', 'Miami']
}
df = pd.DataFrame(data)
```

```python
sorted_df = df.sort_values(by=['City', 'Age'], ascending=False)
sorted_df
```

```python
sorted_index = sorted_df.sort_index()
sorted_index
```
