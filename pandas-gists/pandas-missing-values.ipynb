```python
import pandas as pd

data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'Boris', 'Eugene'],
    'Age': [25, None, 35, 16, 45],
    'City': ['New York', 'Los Angeles', 'Chicago', None, 'Miami']
}
df = pd.DataFrame(data)
```

```python
df.isnull()
```

```python
df_cleared = df.dropna()
df_cleared
```

```python
df_dropped_axis = df.dropna(axis=1)
df_dropped_axis
```

```python
df['Age'] = df['Age'].fillna(df['Age'].mean())
df['City'] = df['City'].fillna('Unknown')
df
```
