```python
import pandas as pd

sales_data = {
    'Region': ['North', 'South', 'North', 'South', 'East', 'North'],
    'Product': ['A', 'A', 'B', 'B', 'A', 'C'],
    'Sales': [100, 80, 120, 90, 70, 110],
    'Month': ['Jan', 'Jan', 'Jan', 'Feb', 'Feb', 'Feb']
}
df = pd.DataFrame(sales_data)
```

```python
grouped = df.groupby('Month').apply(lambda x: x)
grouped
```

```python
grouped = df.groupby('Region')['Sales'].max()
grouped
```

```python
grouped = df.groupby(['Region', 'Product'])['Sales'].sum()
grouped
```

```python
grouped = df.groupby(['Region', 'Product'])['Sales'].agg(['sum', 'mean', 'count'])
grouped
```
