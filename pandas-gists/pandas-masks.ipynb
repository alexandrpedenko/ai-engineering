```python
import pandas as pd

data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'Boris', 'Eugene'],
    'Age': [25, 30, 35, 16, 45],
    'City': ['New York', 'Los Angeles', 'Chicago', 'Boston', 'Miami']
}
df = pd.DataFrame(data)
```

```python
mask = (df['Age'] > 20) & (df['City'] != 'Miami')
df[mask]
```

```python
filtered_df = df[df['Age'].between(20, 35)]
filtered_df
```
